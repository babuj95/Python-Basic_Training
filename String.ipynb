# String

# Replace middle string *


```python
name=input()
stringlength=len(name)
middle=stringlength//2
if stringlength % 2==0:
  name=name[:middle-1]+'**'+name[middle+1:]
  print(name)
else:
  name=name[:middle]+'*'+name[middle+1:]
  print(name)
```

# Print ODD and EVEN Seperately

```python
userInput=input()
a=[]
b=[]
for char in userInput:
  if userInput.index(char)%2==0:
    a.append(char)
  else:
    b.append(char)
print(a)
print(b)
```

```python
userInput=input()
a=[]
b=[]
for index,char  in enumerate(userInput):
  if index%2==0:
    a.append(char)
  else:
    b.append(char)
print(''.join(a))
print(''.join(a),''.join(b))
```
