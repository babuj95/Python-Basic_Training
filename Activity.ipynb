# Employee Salary Calculator

```python
emp_name=input("Enter employee name: ")
emp_basic=int(input("Enter basic salary: "))
HRA=int(input("Enter HRA percentage: "))
DA=int(input("Enter DA percentage: "))
tax=int(input("Enter tax percentage: "))


HRA        =int((emp_basic * HRA) / 100)
DA         =int((emp_basic * DA) / 100)
grossSalary=int( emp_basic + HRA + DA)
Tax        =int((grossSalary * tax) / 100)
NetSalary  =int( grossSalary - Tax)

print("")
print(f"Basic Salary : ₹{emp_basic}")
print(f"HRA          : ₹{HRA}")
print(f"DA           : ₹{DA}")
print(f"Gross Salary : ₹{grossSalary}")
print(f"Tax          : ₹{Tax}")
print(f"Net Salary   : ₹{NetSalary}")
```

# String Indexing Challenge

```python
word = input("Enter a word: ")
print("")
print("First character       : ",word[:1])
print("Last character        : ",word[-1])
print("First three characters: ",word[:3])
print("Last three characters : ",word[-3:])
print("Reverse               : ",word[::-1])
```

# Dynamic Calculator

```python
first_number =int(input("Enter first number : "))
second_number=int(input("Enter second number: "))
print("")
print(f"Addition       : {first_number +  second_number}")
print(f"Subtraction    : {first_number -  second_number}")
print(f"Multiplication : {first_number *  second_number}")
print(f"Division       : {first_number /  second_number}")
print(f"Floor Division : {first_number // second_number}")
print(f"Modulus        : {first_number %  second_number}")
print(f"Power          : {first_number ** second_number}")
```
