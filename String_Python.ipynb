# Palindrom

```python
S=input()
reversed_string=(S[::-1])
if S==reversed_string:
    print('yes')
else:
      print('no')
```

# Vowel Check in String

```python
userInput=input().lower()
vowels='aeiou'
found='false'
for char in userInput:
  if char in vowels:
    found='true'
    break
if found=='true':
  print('yes')
else:
    print("no")
```

```python
userInput=input().lower()
if(any(char in 'aeiou' for char in userInput)):
  print("yes")
else:
  print('no')
```
